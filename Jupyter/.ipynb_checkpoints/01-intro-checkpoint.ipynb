```python
df_asia = df[df['origin']=='Asia']
```

```python
X = df_asia[['model_year', 'vehicle_weight']].head(7).values
X
```

```python
import numpy as np
#XTX = X.T @ X
XTX = np.matmul(X.T, X)
XTX
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
w = np.matmul(np.linalg.inv(XTX), X.T) * y
w
```

```python
sum(sum(w))
```
